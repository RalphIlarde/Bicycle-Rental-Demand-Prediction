The Shapiro-Wilk test was conducted to assess the normality of the numerical variables, including daily bicycle rentals, temperature, feeling temperature, humidity, and wind speed. The results will help determine whether the variables follow a normal distribution and guide the selection of appropriate statistical methods for correlation analysis.

```python

# Import statistical library
from scipy.stats import shapiro
import pandas as pd

# Select numerical variables
normality_vars = ["cnt", "temp", "atemp", "hum", "windspeed"]

# Perform Shapiro-Wilk normality test
normality_results = []

for col in normality_vars:
    stat, p_value = shapiro(df[col].dropna())

    normality_results.append({
        "Variable": col,
        "Shapiro-Wilk Statistic": round(stat, 4),
        "P-Value": round(p_value, 6),
        "Normality Result": (
            "Normal" if p_value > 0.05
            else "Not Normal"
        )
    })

# Display results
normality_df = pd.DataFrame(normality_results)
display(normality_df)
```

The Shapiro-Wilk test revealed that all five numerical variables have p-values below the 0.05 significance level, indicating statistically significant deviations from normality. Therefore, the null hypothesis of normality was rejected for all variables. Based on these findings, Spearman's rank correlation will be used to examine relationships among the variables without assuming normal distributions.

Spearman's rank correlation analysis was conducted to examine the strength and direction of relationships between daily bicycle rental demand and the numerical predictor variables. This method was selected based on the normality test results and is appropriate for identifying monotonic relationships without requiring normally distributed variables.

```python

# Spearman Correlation Analysis

# Select the numerical variables
correlation_vars = ["cnt", "temp", "atemp", "hum", "windspeed"]

# Compute Spearman correlation
spearman_corr = df[correlation_vars].corr(method="spearman")

# Display correlation matrix
display(spearman_corr.round(3))

# Visualize correlation matrix
plt.figure(figsize=(9, 6))

sns.heatmap(
    spearman_corr,
    annot=True,
    fmt=".3f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    center=0
)

plt.title("Spearman Correlation Matrix")
plt.tight_layout()
plt.show()
```

The Spearman correlation analysis revealed strong positive relationships between daily bicycle rentals and both temperature (ρ = 0.622) and feeling temperature (ρ = 0.623), suggesting that higher temperatures are associated with increased rental demand. Meanwhile, humidity (ρ = −0.098) and wind speed (ρ = −0.217) exhibited very weak and weak negative relationships, respectively. Notably, temperature and feeling temperature showed an extremely strong positive correlation (ρ = 0.993), indicating potential multicollinearity that requires further examination before regression modeling.

Variance Inflation Factor (VIF) analysis was conducted to assess multicollinearity among the numerical predictor variables. This helps determine whether strong linear relationships between predictors may affect the stability and interpretation of regression models.

```python

# Multicollinearity Analysis using VIF

from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.tools.tools import add_constant

# Select numerical predictor variables
vif_vars = ["temp", "atemp", "hum", "windspeed"]

X_vif = df[vif_vars].copy()

# Add constant for VIF computation
X_vif = add_constant(X_vif)

# Calculate VIF for each predictor
vif_results = pd.DataFrame({
    "Variable": vif_vars,
    "VIF": [
        variance_inflation_factor(X_vif.values, i)
        for i in range(1, X_vif.shape[1])
    ]
})

# Display results
display(vif_results.round(3))
```

The VIF analysis revealed severe multicollinearity between temperature (VIF = 62.970) and feeling temperature (VIF = 63.632), as both values exceed the commonly accepted threshold of 10. In contrast, humidity (VIF = 1.079) and wind speed (VIF = 1.127) showed no substantial multicollinearity concerns. These findings suggest that retaining both temperature variables may affect coefficient stability, making it advisable to consider removing one during preprocessing.

# Part 3. Perform Data Partitioning

Data partitioning was performed to divide the Bike Sharing Dataset into training and testing sets for machine learning regression modeling. This process allows the models to learn from historical observations and be evaluated using unseen data, helping assess their ability to predict daily bicycle rental demand.


The dataset was partitioned into 80% training and 20% testing sets to evaluate the regression models using unseen observations. Since the dataset contains chronological daily rental records, `shuffle=False` was applied to preserve their temporal order, ensuring that earlier observations were used for training and later observations for testing. This approach better reflects real-world demand forecasting and helps prevent future observations from influencing model training.

```python

# Data Partitioning

from sklearn.model_selection import train_test_split

# Arrange observations chronologically
df = df.sort_values("dteday").reset_index(drop=True)

# Separate explanatory variables (X)
# Exclude identifiers and variables that reveal the target
X = df.drop(columns=[
    "instant",
    "dteday",
    "casual",
    "registered",
    "cnt"
])

# Define target variable (y)
y = df["cnt"]

# Split into 80% training and 20% testing
# Keep chronological order for demand forecasting
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    shuffle=False
)

# Display partitioning results
print("Total Observations:", len(df))
print("Total Predictor Variables:", X.shape[1])

print("\nTraining Set:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\nTesting Set:")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)

print("\nTraining Period:",
      df.loc[X_train.index, "dteday"].min().date(),
      "to",
      df.loc[X_train.index, "dteday"].max().date())

print("Testing Period:",
      df.loc[X_test.index, "dteday"].min().date(),
      "to",
      df.loc[X_test.index, "dteday"].max().date())

print("\nPredictor Variables:")
print(X.columns.tolist())
```

The dataset was successfully divided into 584 training observations (80%) and 147 testing observations (20%), with 11 explanatory variables. The training set covers January 1, 2011, to August 6, 2012, while the testing set covers August 7 to December 31, 2012. This chronological partitioning ensures that the models are trained on earlier records and evaluated on later, unseen observations, providing a realistic assessment of their predictive performance.

### Data Partitioning Summary and Conclusion

Data partitioning was successfully completed using an 80:20 chronological split, separating the explanatory variables from the target variable while excluding identifiers and variables that could cause target leakage. The resulting training and testing sets provide a foundation for evaluating the regression models on unseen observations.

GridSearchCV was used to identify the optimal regularization parameter (`alpha`) through time-series cross-validation. The selected model was then evaluated using RMSE and R² across the training and testing sets to assess its predictive performance and generalization.

```python

# Lasso Regression Training and Testing

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import Lasso
from sklearn.model_selection import GridSearchCV, TimeSeriesSplit
from sklearn.metrics import mean_squared_error, r2_score

# Define numerical and categorical variables
numerical_features = ["temp", "hum", "windspeed"]

categorical_features = [
    "season", "yr", "mnth", "holiday",
    "weekday", "workingday", "weathersit"
]

# Establish preprocessing pipeline
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"),
         categorical_features)
    ]
)

# Establish Lasso pipeline
lasso_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("lasso", Lasso(max_iter=30000))
])

# Define hyperparameter values
lasso_param_grid = {
    "lasso__alpha": [0.01, 0.1, 1, 10, 100, 500, 1000]
}

# Preserve chronological order during cross-validation
tscv = TimeSeriesSplit(n_splits=5)

# Perform hyperparameter tuning
lasso_grid = GridSearchCV(
    estimator=lasso_pipeline,
    param_grid=lasso_param_grid,
    cv=tscv,
    scoring="neg_root_mean_squared_error",
    n_jobs=-1
)

# Train Lasso model
lasso_grid.fit(X_train, y_train)

# Generate predictions
lasso_train_pred = lasso_grid.predict(X_train)
lasso_test_pred = lasso_grid.predict(X_test)

# Evaluate model performance
lasso_cv_rmse = -lasso_grid.best_score_
lasso_train_rmse = np.sqrt(
    mean_squared_error(y_train, lasso_train_pred)
)
lasso_test_rmse = np.sqrt(
    mean_squared_error(y_test, lasso_test_pred)
)

lasso_train_r2 = r2_score(y_train, lasso_train_pred)
lasso_test_r2 = r2_score(y_test, lasso_test_pred)

# Display model results
print("LASSO REGRESSION RESULTS")
print("-" * 40)
print("Best Hyperparameters:", lasso_grid.best_params_)
print(f"CV RMSE: {lasso_cv_rmse:.3f}")
print(f"Train RMSE: {lasso_train_rmse:.3f}")
print(f"Test RMSE: {lasso_test_rmse:.3f}")
print(f"Train R²: {lasso_train_r2:.4f}")
print(f"Test R²: {lasso_test_r2:.4f}")
```

The Lasso Regression model achieved an optimal alpha value of 1, with a cross-validation RMSE of 1,089.884. The model recorded a training RMSE of 680.701 and a testing RMSE of 1,109.814, indicating higher prediction errors on unseen data. Additionally, the training R² of 0.8550 decreased to 0.6495 on the testing set, suggesting possible overfitting or changes in rental patterns over time. Nevertheless, the model explained approximately 64.95% of the variation in daily bicycle rental demand on the testing set, demonstrating useful predictive capability.

## Part 5.4. Machine Learning Regression Tree Model Training and Testing

Decision Tree Regression was applied to predict daily bicycle rental demand by identifying patterns and relationships among the selected predictor variables. Unlike linear regression models, Decision Tree Regression can capture nonlinear relationships through a series of decision rules. The model was evaluated to determine whether this approach could improve predictive accuracy compared with the previously developed regression models.

The Decision Tree Regressor was developed using a preprocessing pipeline that retains the original numerical values and applies one-hot encoding to categorical variables. Numerical scaling was omitted because decision trees are not dependent on feature magnitude. GridSearchCV with time-series cross-validation was used to optimize tree depth and minimum sample requirements, while RMSE and R² were calculated to evaluate training, cross-validation, and testing performance.

```python

# Decision Tree Regression Training and Testing

from sklearn.tree import DecisionTreeRegressor

# Establish preprocessing for tree-based models
tree_preprocessor = ColumnTransformer(
    transformers=[
        ("num", "passthrough", numerical_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"),
         categorical_features)
    ]
)

# Establish Decision Tree pipeline
tree_pipeline = Pipeline([
    ("preprocessor", tree_preprocessor),
    ("decision_tree", DecisionTreeRegressor(random_state=42))
])

# Define hyperparameter values
tree_param_grid = {
    "decision_tree__max_depth": [3, 5, 7, 10, None],
    "decision_tree__min_samples_split": [2, 5, 10],
    "decision_tree__min_samples_leaf": [1, 3, 5]
}

# Perform hyperparameter tuning
tree_grid = GridSearchCV(
    estimator=tree_pipeline,
    param_grid=tree_param_grid,
    cv=tscv,
    scoring="neg_root_mean_squared_error",
    n_jobs=-1
)

# Train Decision Tree model
tree_grid.fit(X_train, y_train)

# Generate predictions
tree_train_pred = tree_grid.predict(X_train)
tree_test_pred = tree_grid.predict(X_test)

# Evaluate model performance
tree_cv_rmse = -tree_grid.best_score_

tree_train_rmse = np.sqrt(
    mean_squared_error(y_train, tree_train_pred)
)

tree_test_rmse = np.sqrt(
    mean_squared_error(y_test, tree_test_pred)
)

tree_train_r2 = r2_score(y_train, tree_train_pred)
tree_test_r2 = r2_score(y_test, tree_test_pred)

# Display model results
print("DECISION TREE REGRESSION RESULTS")
print("-" * 40)
print("Best Hyperparameters:", tree_grid.best_params_)
print(f"CV RMSE: {tree_cv_rmse:.3f}")
print(f"Train RMSE: {tree_train_rmse:.3f}")
print(f"Test RMSE: {tree_test_rmse:.3f}")
print(f"Train R²: {tree_train_r2:.4f}")
print(f"Test R²: {tree_test_r2:.4f}")
```

The Decision Tree Regressor achieved optimal hyperparameters of max depth = 5, minimum samples per leaf = 3, and minimum samples per split = 2, with a cross-validation RMSE of 1,333.710. The training RMSE of 605.782 increased substantially to 1,366.093 on the testing set, while R² declined from 0.8851 to 0.4690. This considerable performance gap suggests possible overfitting or changes in rental patterns over time. Compared with the previous regression models, the Decision Tree demonstrated weaker predictive accuracy on unseen observations, explaining approximately 46.90% of the variation in daily bicycle rental demand.
